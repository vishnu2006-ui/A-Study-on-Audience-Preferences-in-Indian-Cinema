Import the required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import fitz
import re

print("All libraries imported successfully!")
```

Upload your PDF dataset

```python
from google.colab import files

uploaded = files.upload()

pdf_path = list(uploaded.keys())[0]

print("Uploaded file:", pdf_path)
```

Extract text from the PDF

```python
doc = fitz.open(pdf_path)

text = ""

for page in doc:
    text += page.get_text() + "\n"

print("PDF text extracted successfully!")
print(text[:2000])
```

Define the dataset columns

```python
columns = [
    "S.No",
    "Gender",
    "Age",
    "Place",
    "Favourite_Director",
    "Favourite_Hero",
    "Industry_Hit",
    "Underrated_Movie",
    "Overrated_Movie",
    "Classic_Movie",
    "Ultra_Classic_Movie",
    "Disaster_Movie",
    "Favourite_Comedian"
]

print("Dataset columns created!")
```

Extract individual survey responses

```python
lines = [
    line.strip()
    for line in text.splitlines()
    if line.strip()
]

records = []
i = 0

while i < len(lines):
    if re.fullmatch(r"\d+", lines[i]):
        serial_number = int(lines[i])

        if i + 12 < len(lines):
            record = lines[i:i+13]

            if all(re.fullmatch(r"\d+", x) is None
                   for x in record[1:]):
                records.append(record)

        i += 13
    else:
        i += 1

df = pd.DataFrame(records, columns=columns)

print("Data extraction completed!")
print("Number of records:", len(df))
```

Check missing values

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

Clean the dataset

```python
df = df.replace("", np.nan)

df = df.drop_duplicates(subset="S.No")

for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].str.strip()

print("Data cleaning completed!")

df.head()
```

Export the final dataset to Excel

```python
output_file = "TFI_Survey_Analysis.xlsx"

with pd.ExcelWriter(output_file) as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned_Data",
        index=False
    )

    for col in columns[1:]:
        df[col].value_counts().to_frame(
            name="Votes"
        ).to_excel(
            writer,
            sheet_name=col[:31]
        )

print("Excel file created successfully!")
```
